# loc (Lable-based indexing) 
#### syntax df.loc[row-lable,column-lable]
# iloc (index-based indexing) 
#### syntax df.iloc[row-index,column-index]

```python
df # here is lipsis not show all data between index 4 & 9989
```

```python
df.loc[2] # only row without column
```

```python
df.loc[2,"Country"] # give only lable here 2 is lable and country too
```

```python
df.iloc[2,2] # give only index first 2 is row second 2 is column 
```

```python
df.loc[2:10] # dont give 9 or n-1 cause its loc lable not index
```

# Logical operator 
## (|)or (&)and (~)not

```python
df
```

```python
df["Region"].unique() # unique data dont repeated data 
```

```python
df["Region"] == "South" # give bool
```

```python
df[df["Region"] == "South"] # to take out data
```

```python
df['Sales']>200
```

```python
df[df['Sales']>200] # to check data use df[]
```

```python
(df["Region"] == "South") & (df['Sales']>200)
```

```python
df[(df["Region"] == "South") & (df['Sales']>200)]
```

```python
df[((df["Region"] == "South") | (df["Region"] == "West")) & (df["Sales"]> 200)]
```

```python
df[~(df["Region"]=="South")]
```

```python
df[(df["Region"]=="West") & (df["Sub-Category"]=="Chairs") & (df["Sales"]>300)]
```
