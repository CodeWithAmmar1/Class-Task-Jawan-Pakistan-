```python
# 1
# Create a set of numbers and remove duplicates from:
# [1, 2, 2, 3, 4, 4, 5]
# 2
# Find common elements between:
# A = {1, 2, 3, 4} B = {3, 4, 5, 6}
# 3
# Add "orange" and "mango" to an existing set of fruits.
# 4
# Remove "banana" using both remove() and discard().
# 5
# Find the union of:
# X = {10, 20, 30} Y = {30, 40, 50}
numbers=[1, 2, 2, 3, 4, 4, 5]
removeDuplicate = set(numbers)
print(f"Remove duplicate : {removeDuplicate}")
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
print(f"Common : {A.intersection(B)}")
fruits={"apple","grapes","banana"}
fruits.update(["orange","mango"])
print(f"Add fruits {fruits}")
fruits.remove("banana")
print(f"Using Remove {fruits}")
fruits.discard("banana") # discard did'nt give error even the item is not in set()
print(f"Using Discard {fruits}")
X = {10, 20, 30} 
Y = {30, 40, 50}
print(f"Union {X.union(Y)}")
```
