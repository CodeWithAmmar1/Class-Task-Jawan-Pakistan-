# Arbitrary Argument (*) return tuple()


```python
def name(*args): # * means its arbitrary take multiple arguments
    print(args)
name("Ammar","Ali","Arees")

def fruits(*args): 
    print(list(args)) # to convert it in list
fruits("Apple","Graphs","Banana")

def total(*x):
    print(f"Total : {sum(x)}")
total(1,2,3,4,5)

def avg(*temp):
    print(f'Average Temperature : {sum(temp)/len(temp)}')
avg(12,22,65,22)

def item(*items):
    for x in items:
        print(f'Item name: {x}')
item("Apple","Watch","Phone")
```

# Arbitrary Keyword Argument (**) return Dic{}

```python
def emp_detail(**kwargs): # it take argument in dic key value parameter
    print(kwargs)
emp_detail(name="ammar",age=24,city="karachi")
```

```python
def product(**kwargs):
    for key , value in kwargs.items():
        print(f"{key}: {value}")
product(name="dell",price=24000,model="Corei3-4th")
```

```python
def prod_sum(**pro_add):
    print(f"Total : {sum(pro_add.values())}")
prod_sum(apple=44,soap=31,watch=100)
```

```python
#example of positional , Arbitrary Argument , Arbitrary Keyword Argument
def calculate_salary(basic,*allowances, **deductions):
    total_allowance = sum(allowances)
    total_deductions = sum(deductions.values())
    gross_salary = basic + total_allowance
    net_salary = gross_salary - total_deductions

    print('basic salary', basic)
    print('total allowance: ', total_allowance)
    print('total deduction: ', total_deductions)
    print('gross salary:', gross_salary)
    print('net salary: ', net_salary)
calculate_salary(5000, 2000,1000, tax=3000,insurance=200)
```
