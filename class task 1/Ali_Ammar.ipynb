```python
# Remove the extra spaces from both ends of the string " Hello World ".
# Remove spaces only from the left side of " Start now".
# Remove trailing spaces from "Finish line ".
# Replace "apple" with "banana" in the sentence "I have an apple".
# Remove the .jpg suffix from "picture.jpg".
# Remove the "Mr. " prefix from "Mr. John".
# How many times does the letter "e" appear in "experience"?
# Find the position of "@" in the email "user@example.com"
text1 = " Hello World "
text1=text1.strip()
print(f"Remove space: '{text1}'")
text2 = " Start now"
text2=text2.lstrip()
print(f"Remove space left side: '{text2}'")
text3 = "Finish line "
text3=text3.rstrip()
print(f"Remove space right side: '{text3}'")
sentence = "I have an apple"
output=sentence.replace("apple","banana")
print(f"Replace word: {output}")
suffix = "picture.jpg"
suffix = suffix.removesuffix(".jpg")
print(f"Remove suffix: {suffix}")
preffix = "Mr. John"
preffix = preffix.removeprefix("Mr. ")
print(f"Remove preffix: {preffix}")
letter = "experience"
count=letter.count("e")
print(f"Letter e: {count}")
email = "user@example.com"
position=email.find("@")
print(f"Position of @ : {position}")
```
