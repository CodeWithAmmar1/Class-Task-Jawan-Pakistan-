```python
#invoice_id ,track_id ,track_name, media type name ,quantity & unit price where invoice_id==1
query='''
select il.invoice_id,t.track_id,t.name track_name,m.name media_name,il.quantity,il.unit_price
from invoice_line il 
inner join track t
on
il.track_id=t.track_id
inner join media_type m
on
t.media_type_id=m.media_type_id
where il.invoice_id == 1
'''
sql_query(query)
```

```python
#invoice_id ,track_id ,track_name, media type name artist name,,quantity & unit price where invoice_id==1
query='''
select il.invoice_id,t.track_id,t.name track_name,ar.name,m.name media_name,il.quantity,il.unit_price
from invoice_line il 
inner join track t
on
il.track_id=t.track_id
inner join media_type m
on
t.media_type_id=m.media_type_id
inner join album al
on
t.album_id=al.album_id
inner join artist ar
on 
al.artist_id=ar.artist_id

where il.invoice_id == 1
'''
sql_query(query)
```

```python
#track_id,album title,artist name
query='''
select t.track_id,al.title album_title,ar.name artist_name
from track t
inner join album al
on
t.album_id=al.album_id
inner join artist ar
on 
al.artist_id=ar.artist_id

'''
sql_query(query)
```

```python
# count of track ids ,top 10 no. of track album name and artist name
query='''
select count(il.track_id) No_of_track ,al.title album_name,ar.name artist_name
from invoice_line il
inner join track t
on
il.track_id=t.track_id
inner join album al
on
t.album_id=al.album_id
inner join artist ar
on 
al.artist_id=ar.artist_id
group by 2
order by  1 DESC limit 10
'''
sql_query(query)
```

```python
query='''
select c.first_name ||" " || c.last_name customer_name, count(i.invoice_id) No_of_invoice,sum(i.total) total_Spent,
Case
    when sum(i.total)< 40 then "Small Spender"
    when sum(i.total)> 100 then "Big Spender"
    else "Normal Spender"
End as spender_category
from customer c
inner join invoice i
on c.customer_id=i.customer_id
group by c.customer_id
order by 3 DESC 
'''
sql_query(query)
# case for condition
```

```python
# value_count 
q = '''SELECT first_name || ' ' || last_name Customer_Name, 
        count(inv.invoice_id) No_Of_Invoices,
        sum(total) Total_Spent,
        Case
            when sum(inv.total) < 40 then "Small Spender"
            when sum(inv.total) > 100 then "Big Spender"
            Else "Regular"
        End as Customer_category
        from customer c
        inner join invoice inv
        on 
        c.customer_id = inv.customer_id
        group by c.customer_id
        order by 1
        
'''
sql_query(q)["Customer_category"].value_counts()
```

```python
q = """select c.*, sum(inv.total) Total_Spent
        from customer c
        left join invoice inv
        on
        c.customer_id = inv.customer_id
        group by c.customer_id
        having Total_Spent > 90

"""#different between having and where are having write with condition or clause
sql_query(q)
```

```python
# customer all col , Top 3 total invoice
q = """select c.*, sum(inv.total) Total_Spent
        from customer c
        left join invoice inv
        on
        c.customer_id = inv.customer_id
        group by c.customer_id
        having Total_Spent > 90
        order by Total_Spent DESC limit 3
        

"""
sql_query(q)
```

```python
# track_id, total_sale

q = """SELECT t.track_id, sum(il.quantity) Total_Sale FROM track t
        left JOIN invoice_line il
        ON
        t.track_id = il.track_id
        GROUP BY 1
        """

sql_query(q)
```

```python
# create Table
q='''
create table Spender as
select c.first_name ||" " || c.last_name customer_name, count(i.invoice_id) No_of_invoice,sum(i.total) total_Spent,
Case
    when sum(i.total)< 40 then "Small Spender"
    when sum(i.total)> 100 then "Big Spender"
    else "Normal Spender"
End as spender_category
from customer c
inner join invoice i
on c.customer_id=i.customer_id
group by c.customer_id
order by 3 DESC 
'''
conn.execute(q)# save
```

```python
q = """select c.customer_id, c.first_name ||" "|| c.last_name full_name, c.country,
       sum(i.total) Total_purchased
       from customer c
       inner join invoice i 
       on
       c.customer_id = i.customer_id
       group by 1"""
sql_query(q)
```

```python
#  customer from each country that has spent the most money at our store
# Best customer_id, Customer_name, Country, Top_Buyer

q = """with best_customer_id as
       (select c.customer_id, c.first_name ||" "|| c.last_name full_name, c.country,
       sum(i.total) Total_purchased
       from customer c
       inner join invoice i 
       on
       c.customer_id = i.customer_id
       group by 1)
       
       
       select customer_id, country, full_name, max(Total_purchased) Top_buyer
       from best_customer_id
       group by 2

    """
# here we use sub query cause we have to use sum and max at same time which is not possible so we use 
# sub query like make (with best_customer_id) then take sum(i.total) then select from best_customer_id and take max(total_purchased)
sql_query(q)
```
