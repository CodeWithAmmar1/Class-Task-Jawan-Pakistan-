# Class :
#### In object-oriented programming, a class is a blueprint or template used to create objects.
## It defines:
#### Attributes (Data/Variables): What the object has.
#### Methods (Functions/Actions): What the object can do.

```python
class Student :
    def greet(self): #reseverd by obj (self)
        return "Welcome.."
    def greet_with_info(self,name,age): # method "greet" "greet_with_info"
        print("reserverd by object",self)
        self.stu_name=name # attribute
        self.stu_age=age # attribute (stu_age)
        return "Welcome.."
stu=Student() #object (stu) of class
stu.greet() # () cause it was method
```

```python
stu.greet_with_info("ammar",26) # call "print("reserverd by object",self)" it show self reseverd by obj (stu) and return welcome..
```

```python
print(stu.stu_name) # call attribute of method
print(stu.stu_age)
```

```python
class Student :
    def __init__(self,school_name,city,Class): # constructor & init (special method)
        self.school_name=school_name
        self.city_name=city
        self.Class=Class
        # return "welcome.."  # their is no return in init method
    def greet(self): #reseverd by obj (self)
        return "Welcome.."
    def greet_with_info(self,name,age): # method "greet" "greet_with_info"
        print("reserverd by object",self)
        self.stu_name=name # attribute
        self.stu_age=age # attribute (stu_age)
        return "Welcome.."
        
# now we make object 
# before...  we make obj , when we call method at that time we pass parameter
# but now our class has init method or constructor so we have to pass parameter when we are make obj

stud=Student("WHGS","Karachi","10th") # as we know our constructor has init method and some parameter
print(stud.city_name)
print(stud.school_name)
print(stud.Class)
```
