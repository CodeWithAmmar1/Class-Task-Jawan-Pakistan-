```python
class Student :
    def __init__(self,school_name,city,Class): # constructor & init (special method)
        self.school_name=school_name
        self.city_name=city
        self.Class=Class
        # return "welcome.."  # their is no return in init method
    def greet(self): # method
        return "Welcome.."
    def greet_with_info(self,name,age): # method 
        print("reserverd by object",self)
        self.stu_name=name # attribute
        self.stu_age=age # attribute
        return "Welcome.."
        
# now we make object 
# before...  we make obj , when we call method at that time we pass parameter
# but now our class has init method or constructor so we have to pass parameter when we are make obj

stud=Student("WHGS","Karachi","10th") # as we know our constructor has init method and some parameter
print(stud.city_name)
print(stud.school_name)
print(stud.Class)
```

# ***Inheritance***
Inheritance is a fundamental concept in object oriented programming (OOP) where a new class, called the child class or subclass, is derived from an existing class, known as the parent class or superclass. The child class inherits the attributes and methods of the parent class, allowing for code reusability and the ability to extend or modify the behavior of the parent class without altering its original code.

**Key Points:**

* Purpose: To promote reusability, maintainability, and scalability of code.
* Access: The child class can use or override the methods and properties of the parent class.

```python
class Employee:
    def __init__(self,name,Employee_id,salary): # constructor
        self.name = name # attribute
        self.Employee_id = Employee_id
        self.salary = salary
    def work(self):
        print(f'{self.name} is working') # inherite attribute of super Class/parent Class
    def get_details(self):
        print(f"Name: {self.name}, Employee id: {self.Employee_id}, Salary: {self.salary}")

employee = Employee('Ammar',103,50000) #object
print(employee.name) # attribute
employee.get_details() # calling method
```

```python
class Manager (Employee): # Child class Employee pass from parent (make same parameter in __init__ of child same as parent
    def __init__(self,name,Employee_id,salary,email,team_size):
        self.name = name # First give parent attribute info
        self.Employee_id = Employee_id
        self.salary = salary
        self.email = email # then make addition if want
        self.team_size = team_size

employee = Manager('Ammar',100,90000,"aliammar@gmail.com",20)
print(employee.Employee_id)
employee.get_details()
```

```python
class ManagerCallSuper(Employee): # pass parent Employee
    def __init__(self,name,Employee_id,salary,email,team_size): # What if their is 100 of parameter in parent
        super().__init__(name,Employee_id,salary) # auto way save time and line use super().__init__ to avoid self.xy
        self.email = email
        self.team_size = team_size

employee = ManagerCallSuper('Ammar',100,90000,"aliammar@gmail.com",20)
print(employee.salary)
```

```python
class Developer (Employee): # inherite parent
    def __init__(self,name,Employee_id,salary,email,team_size):
        self.name = name
        self.Employee_id = Employee_id
        self.salary = salary
        self.email = email
        self.team_size = team_size
    def code(self):
        print(f"Team Size : {self.team_size}")

dev = Developer('Ammar',100,90000,"aliammar@gmail.com",20)
print(dev.email)
dev.code()
```

# ATM Example

```python
class ATM:
    def withdraw(self, amount, currency="USD"):
        if currency == "USD":
            print(f"Withdrawing {amount} dollars.")
        elif currency == "INR":
            print(f"Withdrawing {amount} INR rupees.")
        elif currency == "PKR":
            print(f"Withdrawing {amount} rupees.")
        else:
            print("Currency not supported.")
```

```python
atm = ATM() #object
```

```python
atm.withdraw(100) # call method
```

```python
atm.withdraw(5000, "INR")
```

```python
atm.withdraw(10000, "PKR")
```

```python
atm.withdraw(100, "USDT")
```

## ***Overriding***

```python
class greet1():
    def welcome(self):
        print("Welcome to class 1")
class greet2(greet1): # inherite
    def welcome(self):
        print("Welcome to class 2")
obj=greet2()
obj.welcome() # override welcome method both have same name of method so it override class 1 print
```

```python
class greet1():
    def welcome(self):
        print("Welcome to class 1")
class greet2(greet1):
    def welcome(self):
        super().welcome() # call method in method 2
        print("Welcome to class 2")

obj=greet2()
obj.welcome() # override welcome method print both now cause we call method 2 in method 2
```

## ***Payment Processing System***

```python
class Payment:
    def process_payment(self, amount):
        print(f"Processing payment of {amount}.")

class CreditCardPayment(Payment):
    def process_payment(self, amount):
        super().process_payment(amount) # overriding method
        print(f"Processing credit card payment of {amount}.")
```

```python
pay = Payment() #obj
```

```python
pay.process_payment(100)
```

```python
credit_card_payment = CreditCardPayment()
```

```python
credit_card_payment.process_payment(2000)
```

## ***Abstraction***

Hiding the implementation details of a class and only showing the essential features to the users.

```python
class Car:
    def __init__(self,model, year):
        self.model = model
        self.year = year

    def start(self):
        self.clutch = True # here this info is hide that clutch is true
        print("Starting the car")

    def accelerate(self):
        self.acc = True # same here
        print("Accelerating")

    def brake(self):
        print("Braking")

    def steer(self):
        print("Steering...")
```

```python
car = Car("Toyota", 2023)
car.start() # see their is no clutch shown
```

```python
car.clutch # even we check it but when we call method it hide
```

## ***Encapsulation***

**Encapsulation is a fundamental principle in object oriented programming that focuses on bundling data and the methods that operate on that data into a single unit called a class. It allows you to control the access and visibility of the data and methods, providing a way to protect and organize your code**



## ***Access Modifiers***
**Public:** Accessible from anywhere in the program.

**Private:** Restricted access, typically indicated by a double underscore (__) prefix. Private members are not directly accessible outside the class.

**Protected:** Indicated by a single underscore (_), and are meant to be accessed within the class and its subclasses.

```python
class Account:
    def __init__(self, username, password):
        self.__username = username  # Private attribute
        self.__password = password  # Private attribute

    # Getter for username
    def get_username(self):
        return self.__username

    # Setter for username
    def set_username(self, new_username):
        self.__username = new_username

    # Getter for password
    def get_password(self):
        return self.__password

    # Setter for password
    def set_password(self, new_password):
        self.__password = new_password


account1 = Account("sara", "1234")
```

```python
account1.__username # not accessable cause it was private
```

```python
account1.username # not even accessable if we remove __
```

```python
account1.get_username()
```

```python
account1.get_password()
```

```python
account1.set_username('Ammar')
```

```python
account1.get_username()
```

```python
account1.set_password("3456")
```

```python
account1.get_password()
```

```python
class Parent:
    def __init__(self, name):
        self.__name = name  # Private attribute

class Child(Parent): # inherite in child class (private attribute) which is not accessable 
    def print_name(self):
        # Trying to access private attribute
        print(self.__name)

child = Child("Sara")
child.print_name()  # This will raise an AttributeError
```

```python
class Person:
    def __init__(self,name,age):
        self._name=name # protected attribute
        self._age=age
```

```python
person=Person("ammar",32) 
```

```python
person._name # Technically can access outside, but not recommended
```

```python
class Student(Person): # inherite in child (protected attribute)
    def __init__(self,name,age):
        super().__init__(name,age)

    def display_info(self): 
        print(f"the person name is {self._name} and the age is {self._age}")
```

```python
student1=Student("ammar",25)
```

```python
student1.display_info()
```

```python
student1._name
```

```python
student1._age
```
