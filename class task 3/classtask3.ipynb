```python
# practice
#Write a program to print countdown reverse order if user input random number

# Example
# Enter number: 5
# Output:
# 5
# 4
# 3
# 2
# 1
# Time up
num=int(input("Enter number to start countdown ..."))
print("*"*20)
print("Countdown Start ...")
print("*"*20)
for x in range(num,0,-1):
    print(x)
print("Time up")
print("*"*20)
```

```python
#Write a Python program to calculate the sum of numbers from 1 to 10 using a for loop.
# (1+2+3....+10)
# sum = 55
result=0
for x in range(1,11):
    result+=x
print(f"sum = {result}")
```

```python
#Write a Python program to count how many vowels are present in a given string using a for loop.
vowels="aeiou"
count=0
text = input("Enter a string: ").lower()
for x in text:
    if x in vowels:
        count+=1
print(f"Count : {count}")
```

```python
# Write a Python program to print the multiplication table of a given number using a while loop.
x=1
number=int(input("Enter a number to print table :"))
while x <= 10:
    print(f"{number} x {x} = {x*number}")
    x+=1
```
