# Error Handling (Try Execpt)

```python
print(1/0)  # it give error cause nothing is divided by zero 
print("This line isnt work")
```

```python
try : 
    print(1/0)
except ZeroDivisionError: # for specific Error (Should define exect error)
    print("Cannot divide by Zero...")
print("This line now work")
    
```

```python
try:
    print(1/"y")
except TypeError: # specific error
    print("Type Error")
```

```python
try:
    num1=int(input("Enter First Number"))
    num2=int(input("Enter Second Number"))
    result=num1/num2
    print("(try) Division : ",result) # print result here or in else same thing
except ZeroDivisionError:
    print("Cannot divide by Zero...")
except: # default error take all error not need to define( must be in last of every error)
    print("Error")
else:# same print line in try
    print("(else) Division : ",result) 
finally: # program run or not this line will run
    print("Program Executed...") 
```
