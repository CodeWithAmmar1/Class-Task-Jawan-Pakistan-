```python
# we have to drop it
df.drop(["Vehicle Size"],axis=1,inplace=True)
```

```python
plt.title("Numbers of cars by make")
plt.xlabel("Make")
plt.ylabel("Number of car")
plt.show()
```

```python
df.Make.value_counts().plot(kind="bar")
plt.title("Numbers of cars by make")
plt.xlabel("Make")
plt.ylabel("Number of car")
plt.show()
```

```python
# top 10
df.Make.value_counts().head(10).plot(kind="bar")
plt.title("Numbers of cars by make")
plt.xlabel("Make")
plt.ylabel("Number of car")
plt.show()
```

```python
# bottom 10
df.Make.value_counts().tail(10).plot(kind="bar")
plt.title("Numbers of cars by make")
plt.xlabel("Make")
plt.ylabel("Number of car")
plt.show()
```
